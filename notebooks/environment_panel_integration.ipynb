### Preprocessing

#### gdf munis

```python
gdf_munis.drop('CVE_MUN', axis=1, inplace=True)
print(gdf_munis.columns)
```

```python
gdf_munis.rename(columns={'CVEGEO': 'muni_id', 'CVE_ENT': 'state_id', 'NOMGEO': 'municipality'}, inplace=True)
gdf_munis = gdf_munis[gdf_munis.state_id.isin(["12", "16", "15"])].reset_index(drop=True)

gdf_munis['muni_id'] = gdf_munis['muni_id'].astype('str')
gdf_munis['state_id'] = gdf_munis['state_id'].astype('int64')
gdf_munis['municipality'] = gdf_munis['municipality'].astype('str')


print(gdf_munis.shape)
gdf_munis.head()
```

#### Population by municipality

```python
df_pop_m.columns = df_pop_m.columns.str.lower()
df_pop_m.rename(columns={'ano': 'year', 'clave_ent': 'state_id', 'nom_ent': 'state', 'clave': 'muni_id', 'nom_mun': 'municipality', 'sexo': 'sex', 'pob_total': 'pop_total'}, 
                  inplace=True)

df_pop_m = df_pop_m[df_pop_m.state_id.isin([12, 16, 15])]
df_pop_m = df_pop_m[df_pop_m.year == 2025].reset_index(drop=True)

print(df_pop_m.shape)
df_pop_m.head()
```

```python
# Get mean and median of population across all municipalities

print(f'Mean across all municipalities: {df_pop.pop_total.mean()}')
print(f'Median across all municipalities: {df_pop.pop_total.median()}')
```

## Create adjacency dataframe

```python
# Create adjacency


sindex = gdf_munis.sindex
edges = []

for idx, row in gdf_munis.iterrows():
    geom = row.geometry
    muni_i = row["muni_id"]

    # candidates from bounding box
    candidates = list(sindex.intersection(geom.bounds))

    for cand in candidates:
        if cand <= idx:
            continue
        
        geom_j = gdf_munis.loc[cand, "geometry"]
        muni_j = gdf_munis.loc[cand, "muni_id"]

        # check if polygons actually touch
        if geom.touches(geom_j) or geom.intersects(geom_j):
            edges.append((muni_i, muni_j))

edges_df = pd.DataFrame(edges, columns=["source", "target"])
```

```python
print(edges_df.shape)
edges_df.head()
```

```python
# Map the node id to the edges data frame

id_map = dict(zip(df_munis["muni_id"], df_munis["node_id"], strict=False))

edges_df["source"] = edges_df["source"].map(id_map)
edges_df["target"] = edges_df["target"].map(id_map)
```

```python
edges_df = edges_df.dropna(subset=["source", "target"])
print(edges_df.shape)
```
